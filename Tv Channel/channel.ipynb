```python
df['Channel_Name'] = df['Channel_Name'].str.strip()
df['Channel_Name'] = df['Channel_Name'].str.title()
```

```python
numeric_columns = [
    'Avg_Viewers',
    'Total_Likes',
    'Social_Media_Followers',
    'YouTube_Views',
    'Monthly_Reach'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.dtypes
```

```python
df.isnull().sum()
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

categorical_columns = [
    'Channel_Name',
    'Language',
    'Channel_Category',
    'Popular_Show',
    'Show_Category',
    'HD_Available'
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df.loc[
    (df['Avg_Rating'] < 0) | (df['Avg_Rating'] > 5),
    'Avg_Rating'
] = np.nan
df['Avg_Rating'] = df['Avg_Rating'].fillna(df['Avg_Rating'].median())
```

```python
df.loc[df['Subscription_Fee'] < 0, 'Subscription_Fee'] = np.nan

df['Subscription_Fee'] = df['Subscription_Fee'].fillna(
    df['Subscription_Fee'].median()
)
```

```python
df.loc[df['Years_Active'] < 0, 'Years_Active'] = np.nan

df['Years_Active'] = df['Years_Active'].fillna(
    df['Years_Active'].median()
)
```

```python
df = df.replace(
    ['NA', 'N/A', 'Unknown', 'unknown', ''],
    np.nan
)
```
