```python
import pandas as pd
df=pd.read_csv("mcu.csv")
print(df)
```

```python
df['movie_duration']=df['movie_duration'].fillna(df['movie_duration'].mean)
print(df['movie_duration'])
df['production_budget']=df['production_budget'].fillna(df['production_budget'].mean)
print(df['production_budget'])
df['opening_weekend']=df['opening_weekend'].fillna(df['opening_weekend'].mean)
print(df['opening_weekend'])
df['domestic_box_office']=df['domestic_box_office'].fillna(df['domestic_box_office'].mean)
print(df['domestic_box_office'])
df['worldwide_box_office']=df['worldwide_box_office'].fillna(df['worldwide_box_office'].mean)
print(df['worldwide_box_office'])
```

```python
print("\nmissing value after cleaning:")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt

plt.plot(df["movie_title"], df["tomato_meter"], marker="*")

plt.title("Tomato Meter of Movies")
plt.xlabel("Movie Title")
plt.ylabel("Tomato Meter")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.bar(
    ["Tomato Meter", "Audience Score"],
    [df["tomato_meter"].mean(), df["audience_score"].mean()]
)

plt.title("Average Movie Scores")
plt.xlabel("Score Type")
plt.ylabel("Average Score")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["tomato_meter"], df["audience_score"])

plt.title("Tomato Meter vs Audience Score")
plt.xlabel("Tomato Meter")
plt.ylabel("Audience Score")
plt.show()
```
